```python
import numpy as np
import matplotlib.pyplot as plt
t = np.linspace(0, 2*np.pi, 400)
x, y = np.cos(t), np.sin(t)
plt.figure(figsize=(4,4)); plt.plot(x, y)
plt.axis('equal'); plt.title('Hola Robótica'); plt.show()
```
